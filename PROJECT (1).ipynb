# 3. Univeriate Analysis

# 3.1 . Numerical variables

```python
df=pd.read_csv("bank-additional-full.csv",sep=";")
num_cols=df.select_dtypes(include=np.number).columns

# descriptive stats
df[num_cols].describe

# Histograms
df[num_cols].hist(figsize=(12,10))
plt.show()

# Boxplots(outliers)
for col in num_cols:
    plt.figure()
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

# Outlier Detection (IQR)

```python
def detect_outlier(col):
    Q1=col.quantile(0.25)
    Q3=col.quantile(0.75)
    IQR=Q3-Q1
    return col[(col<Q1-1.5*IQR)| (col>Q3+1.5*IQR)]

for col in num_cols:
    outliers=detect_outlier(df[col])
    print(col, ":",len(outliers))
```

# 3.2 Catergorical variables

```python
cat_cols=df.select_dtypes(include='object').columns

for col in cat_cols:
    print(f"/n{col}")
    print(df[col].value_counts())
    print(df[col].value_counts(normalize=True)*100)

    sns.countplot(y=col, data=df,order=df[col].value_counts().index)
    plt.title(col)
    plt.show()
```

# 4.Bivarite Analysis

# 4.1 Target vs Categorical features

```python
df=pd.read_csv("bank-additional-full.csv",sep=";")

cat_cols=df.select_dtypes(include='object').columns

for col in cat_cols:
    if col!='y':
        ctab=pd.crosstab(df[col], df['y'], normalize='index')*100
        print(ctab)

        ctab.plot(kind='bar', stacked=True)
        plt.title(f"{col} vs Target")
        plt.ylabel("percentage")
        plt.show()
```

# Correlation with target

```python
# convert target to numeric
df=pd.read_csv("bank-additional-full.csv",sep=";")
df['y_num']=df['y'].map({'no':0, 'yes':1})

correlations=df.corr(numeric_only=True)['y_num'].sort_values(ascending=False)
print(correlations)
```

# 5.Multivariate Analysis

# Highly Corralated Features

```python
corr_matrix=df.corr(numeric_only=True).abs()
high_corr=corr_matrix[(corr_matrix > 0.7) & (corr_matrix < 1)]
print(high_corr)
```

# 5.1 Pivot Tables (Interaction Effects)

```python
pivot=pd.pivot_table(df,values='y_num',
                     index='job',
                     columns='education',
                     aggfunc=np.mean)
print(pivot)

sns.heatmap(pivot,annot=True, cmap='coolwarm')
plt.title("job vs education vs Subscription Rate")
plt.show()
```
